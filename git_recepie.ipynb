بسیار عالی. چون ریپازیتوری از قبل وجود دارد، ما باید آن را **یک‌بار برای همیشه** داخل گوگل درایو شما «کلون» (Clone) کنیم. بعد از آن، فایل‌ها همیشه در درایو می‌مانند و شما فقط نوت‌بوک‌ها را باز می‌کنید و کد می‌زنید.

این مراحل را قدم‌به‌قدم انجام دهید:

### ۱. ساخت توکن دسترسی (GitHub Token)
چون گیت‌هاب دیگر رمز عبور معمولی را برای ترمینال قبول نمی‌کند، باید یک Token بسازید (اگر قبلاً نساخته‌اید):
1. در گیت‌هاب بروید به: **Settings** -> **Developer settings** -> **Personal access tokens** -> **Tokens (classic)**.
2. دکمه **Generate new token** را بزنید.
3. تیک گزینه‌ی **repo** (برای دسترسی کامل به ریپازیتوری‌ها) را بزنید.
4. توکن را کپی کنید و در یک جای امن نگه دارید (چون دیگر آن را نشان نمی‌دهد).

---

### ۲. اتصال کولب به درایو و تنظیمات اولیه
یک نوت‌بوک جدید در کولب باز کنید و این کدها را اجرا کنید تا درایو وصل شود:

```python
from google.colab import drive
drive.mount('/content/drive')
```

حالا باید به پوشه‌ای در درایو برویم که می‌خواهید پروژه آنجا باشد (مثلاً پوشه `Colab Notebooks`):

```python
%cd /content/drive/My Drive/Colab Notebooks/
```

---

### ۳. کلون کردن ریپازیتوری موجود (فقط یک‌بار)
حالا با استفاده از آن توکن که ساختید، ریپازیتوری را دانلود می‌کنیم.
*(در کد زیر جای `YOUR_TOKEN` توکن خودتان، جای `USERNAME` نام کاربری و جای `REPO_NAME` اسم ریپازیتوری خودتان را بگذارید)*:

```bash
!git clone https://YOUR_TOKEN@github.com/USERNAME/REPO_NAME.git
```

✅ **تمام شد!** الان کل پروژه شما داخل گوگل درایو ذخیره شده است.

---

### ۴. روش کار از این به بعد (روزمره)
از امروز به بعد، هر وقت خواستید روی تمرین جدید کار کنید:

1. **باز کردن فایل:** در گوگل کولب، از منوی سمت چپ (آیکون پوشه)، به مسیر `drive/MyDrive/Colab Notebooks/REPO_NAME` بروید و فایل نوت‌بوک (.ipynb) مورد نظرتان را باز کنید.
2. **شروع کار روی برنچ جدید:**
   داخل همان نوت‌بوک، یک سلول کد باز کنید و دستورات زیر را بزنید تا وارد پوشه پروژه شوید و برنچ بسازید:

   ```python
   # اول وارد پوشه پروژه شوید
   %cd /content/drive/My Drive/Colab Notebooks/REPO_NAME
   ```

   ```bash
   # تنظیم ایمیل و نام (اگر سشن جدید است)
   !git config --global user.email "your_email@example.com"
   !git config --global user.name "Your Name"

   # ساخت برنچ برای تمرین جدید
   !git checkout -b hw2-neural-networks
   ```

3. **کد زدن و ذخیره:** کد بزنید و تغییرات را با `Ctrl+S` در کولب ذخیره کنید.

4. **کامیت و پوش کردن:**
   هر وقت خواستید تغییرات را به گیت‌هاب بفرستید:

   ```bash
   !git add .
   !git commit -m "Complete section 1 of HW2"
   !git push origin hw2-neural-networks
   ```
   *(نکته: چون با توکن کلون کردید، موقع پوش کردن دیگر رمز نمی‌خواهد).*

5. **مرج کردن (پایان کار):**
   ```bash
   !git checkout main
   !git pull origin main
   !git merge hw2-neural-networks
   !git push origin main
   ```

**خلاصه:**
فقط مرحله ۳ را الان انجام دهید. از این به بعد فایل‌ها در درایو شما هستند و مثل یک پوشه معمولی با آن‌ها رفتار کنید. هر سوالی در مورد خطاهای احتمالی گیت داشتید بپرسید.

````python
import os
import sys
from google.colab import drive
import getpass
import subprocess

# --- 1. تنظیمات مسیر پروژه ---
# مسیر دقیق پوشه پروژه شما در گوگل درایو
PROJECT_PATH = "/content/drive/MyDrive/Colab Notebooks/Master_course_Homeworks"

print("🚀 Starting Git Setup for Google Colab...")

# --- 2. اتصال به گوگل درایو (Mount) ---
if not os.path.exists('/content/drive'):
    print("📂 Mounting Google Drive...")
    drive.mount('/content/drive')
else:
    print("✅ Google Drive is already mounted.")

# --- 3. رفتن به پوشه پروژه ---
try:
    os.chdir(PROJECT_PATH)
    print(f"📂 Changed directory to: {os.getcwd()}")
except FileNotFoundError:
    print(f"❌ Error: Directory not found: {PROJECT_PATH}")
    print("Please check the path in your Google Drive.")
    sys.exit(1)

# --- 4. بررسی وضعیت گیت ---
if not os.path.exists(".git"):
    print("❌ Error: This is not a git repository (no .git folder found).")
    print("Did you initialize git here?")
    sys.exit(1)

print("\n" + "="*40)
print("🔐 GitHub Authentication")
print("We need your GitHub Username and Personal Access Token (PAT).")
print("If you don't have a PAT, go to: https://github.com/settings/tokens")
print("Select 'Generate new token (classic)' -> Check 'repo' scope -> Generate")
print("="*40 + "\n")

# --- 5. دریافت اطلاعات کاربری ---
# یوزرنیم پیش‌فرض شما برای راحتی کار
DEFAULT_USERNAME = "ArsHia-cdMstr"
username = input(f"GitHub Username (Press Enter for '{DEFAULT_USERNAME}'): ").strip()
if not username:
    username = DEFAULT_USERNAME

password = getpass.getpass("🔑 Paste your GitHub PAT (Hidden): ").strip()

if not password:
    print("❌ Error: Password/Token cannot be empty.")
    sys.exit(1)

# --- 6. تنظیم ریموت و پوش کردن ---
repo_url = f"https://{username}:{password}@github.com/{username}/Master_course_Homeworks.git"

print("\n🔄 Updating Remote URL...")
subprocess.run(["git", "remote", "remove", "origin"], capture_output=True)
subprocess.run(["git", "remote", "add", "origin", repo_url], check=True)

print("\n📤 Pushing to GitHub (Branch: proj-MAS)...")
try:
    # اول سعی می‌کنیم پول کنیم تا اگر تغییری هست بگیریم (برای جلوگیری از تداخل)
    # subprocess.run(["git", "pull", "origin", "proj-MAS"], capture_output=True)

    # پوش کردن روی برنچ proj-MAS
    result = subprocess.run(["git", "push", "-u", "origin", "proj-MAS"],
                            capture_output=True, text=True)

    if result.returncode == 0:
        print("\n✅ SUCCESS! Project pushed successfully.")
        print(result.stdout)
    else:
        print("\n❌ PUSH FAILED!")
        print("Git Output:")
        print(result.stderr)

except Exception as e:
    print(f"\n❌ An error occurred: {e}")

print("\n🏁 Done.")
````
